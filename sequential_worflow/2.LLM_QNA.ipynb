```python
from langgraph.graph import StateGraph,START,END
from dotenv import load_dotenv
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
```

```python
class QAState(TypedDict):
    qns:str
    ans:str
```

```python
def LlmQa(state:QAState):
    qn = state["qns"]
    prompt = f"Answer the user question {qn}"
    ans = llm.invoke(prompt).content
    state["ans"] = ans
    return {"ans":ans}
    
```

```python


graph = StateGraph(QAState)


graph.add_node("llmqa",LlmQa)
```

```python
graph.add_edge(START,"llmqa")
graph.add_edge("llmqa",END)
```

```python
workflow = graph.compile()
```

```python
results = workflow.invoke({"qns":"How to become super deadly in our field foe ex software enginering"})
print(results["ans"])
```
