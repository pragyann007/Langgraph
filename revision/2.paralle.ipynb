```python
from langgraph.graph import START,END,StateGraph
from typing import TypedDict
```

```python
class IntVals(TypedDict):
    ins:str
    first:str
    second:str
    merged:str

```

```python
def Ifirst(state:IntVals):
    vals = f"a {state["ins"]}"
    state["first"]= vals
    return {"first":vals}

def ISecond(state:IntVals):
    vals = f"b {state["ins"]}"
    state["second"]=vals
    return {"second":vals}
def IMerge(state:IntVals):
    vals = f"abc {state["second"]}"
    state["merged"]=vals
    return {"merged":vals}

```

```python
graph = StateGraph(IntVals)
graph.add_node("a",Ifirst)
graph.add_node("b",ISecond)
graph.add_node("c",IMerge)


graph.add_edge(START,"a")
graph.add_edge(START,"b")
graph.add_edge("a","c")
graph.add_edge("b","c")
graph.add_edge("c",END)
```

```python
workflow=graph.compile()
```

```python
final = workflow.invoke({"ins":"helloworld"})
print(final)
```
