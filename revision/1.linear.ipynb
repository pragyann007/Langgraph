```python
from langgraph.graph import START,END,StateGraph
from typing import TypedDict
```

```python
class BMIState(TypedDict):
    weight:float
    height:float
    bmi:float
    
```

```python
def bmi_calculator(state:BMIState):
    bmi = state["weight"]/(state["height"]**2)

    state["bmi"]=bmi 
    return {"bmi":bmi}
```

```python
graph = StateGraph(BMIState)
```

```python
graph.add_node("bmi_calculator",bmi_calculator)
```

```python
graph.add_edge(START,"bmi_calculator")
graph.add_edge("bmi_calculator",END)
```

```python
workflow = graph.compile()
```

```python
initial = {
    "weight":67,
    "height":1.3
}

final = workflow.invoke(initial)

print(final)
```
